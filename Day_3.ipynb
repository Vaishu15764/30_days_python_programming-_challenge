## conditional statetments

**1. check  number is positive , negative or zero**

```python
num=eval(input("enter the number:"))
if num>0:
    print(f'{num} is positive number')
elif num<0:
    print(f'{num} is negative number')
else:
    print('zero')
```

**2. check even or odd**

```python
num=eval(input('enter the number:'))
if num%2==0:
    print('even number')
else:
    print('odd number')
```

**4.find greater of two numbers**

```python
a=eval(input('enter a:'))
b=eval(input('enter b:'))
if a>b:
    print('a is grater')
elif b>a:
    print('b is greater')
else:
    print('both are equal')
```

**5.find greatest of three numbers**

```python
a=eval(input('enter a:'))
b=eval(input('enter b:'))
c=eval(input('enter c:'))
if a>b and a>c:
    print('a is greater')
elif b>a and b>c:
    print('b is greater')
elif c>a and c>b:
    print('c is greater')
else:
    print('all are equal')
    
```

**9.Check whether a character is a vowel or consonant.**

```python
char=input('enter character:').lower()
v='aeiou'
if len(char)!=1:
    print('please enter only one character')
elif char in v:
    print(f'{char} is vowel')
elif char.isalpha() and char not in v:
    print(f'{char} is consonants')
else:
    print('invalid input')
```

**10. Check whether a number is a single-, double-, or triple-digit number.**

```python
num=abs(eval(input('enter number:')))
if num<=9:
    print('single digit number ')
elif num<=99:
    print('double digit number')
elif num<=999:
    print('triple digit number')
else:
    print('not single , double or triple number')
```

**13.Electricity bill calculator**

```python
units=eval(input('enter units used:'))
if units<=100:
    bill=units*5
elif units<=200:
    bill=100*5 + ((units-100)*7)
elif units<=300:
    bill=100*5 + 100*7 + ((units-200)*10)
else:
    bill=100*5 + 100*7 + 100*10 + ((units-300)*15)
print('total bill =',bill)
```

**14. simple calulator**

```python
print('**Simple Calculator**')
num1=float(input('enter first number:'))
num2=float(input('enter second number:'))
operation=input('enter which opration you want to perform:')

if operation=='+':
    add=num1+num2
    print(f'{num1} + {num2} = {add}')
elif operation=='-':
    sub=num1-num2
    print(f'{num1} - {num2} = {sub}')
elif operation=='*':
    print(f'{num1} * {num2} = {num1*num2}')
elif operation=='/':
    if num2!=0:
        print(f'{num1} / {num2} = {num1/num2}')
    else:
        print('cannot divide by zero')
else:
    print('please enter valid operation')
```

**15. Day number → day name**

**Input 1–7 and print Monday–Sunday.**

```python
day_num=int(input('Enter the day number (1-7):'))
if day_num==1:
    print('Monday')
elif day_num==2:
    print('Tuesday')
elif day_num==3:
    print('Wednesday')
elif day_num==4:
    print('Thursday')
elif day_num==5:
    print('Friday')
elif day_num==6:
    print('Saturday')
elif day_num==7:
    print('Sunday')
else:
    print('please enter valid day number')
```
