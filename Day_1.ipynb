**8. Swap two numbers.**

- using temp 

```python
a= 10
b=20
print(f'Before swap: a={a} , b={b}') 
temp=a
a=b
b=temp
print(f'After swap: a={a} , b={b}')
```

- without temp 

```python
a=10
b=20
print(f'Before swap: a={a} , b={b}')
a,b=b,a
print(f'After swap: a={a} , b={b}')
```
