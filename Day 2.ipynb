**11. check leap year**`

```python
year= int(input('Enter year:'))
if (year%4==0) or (year%100!=0) and (year%400==0):
    print('leap year')
else:
    print('not leap year')
```

**12. simple calculator**

```python
print('Simple Calculator')
print('Menu:')
print('0.Addition \n1.Substraction \n2.Multiplication \n3.Division') 
num=int(input('Enter choice (0-3):'))
a=eval(input('enter first number:'))
b=eval(input('enter second number:'))

if num==0:
    print(f'Adddition \n {a} + {b} = {a+b}')
elif num==1:
    print(f'Substraction \n {a} -{b} = {a-b}')
elif num==2:
    print(f'Multiplication \n {a} x {b} = {a*b}')
elif num==3:
    if b!=0:
        print(f'Division \n {a} / {b} = {a/b}')
    else:
        print('Division by zero is not allowed')
else:
    print('Please enter valid choice')
```

**18. calculate SI (simple interest) and CI (compound interest)**

```python
p= eval(input('enter principal:'))
r= eval(input('enter rate:'))
t= eval(input('enter time:'))

# simple interest
SI= (p*r*t)/100
print(f'Simple Interest= {SI}')

#compound interest
Amount= p*(1+r/100)**t
print(f'Amount= {Amount}')
CI= Amount-p
print(f'Compound Interest= {CI}')
```
